```python
if img_rgb.shape[0] >= img_rgb.shape[1]:
  if (img_rgb.shape[0] - img_rgb.shape[1]) % 2 == 1:
    trimy = int((img_rgb.shape[0] - img_rgb.shape[1] - 1)/2)
    x1 = 0
    x2 = img_rgb.shape[1]
    y1 = trimy
    y2 = img_rgb.shape[0] - trimy - 1
  else:
    trimy = int((img_rgb.shape[0] - img_rgb.shape[1])/2)
    x1 = 0
    x2 = img_rgb.shape[1]
    y1 = trimy
    y2 = img_rgb.shape[0] - trimy
else:
  if (img_rgb.shape[1] - img_rgb.shape[0]) % 2 == 1:
    trimx = int((img_rgb.shape[1] - img_rgb.shape[0] - 1) / 2)
    x1 = trimx
    x2 = img_rgb.shape[1] - trimx - 1
    y1 = 0
    y2 = img_rgb.shape[0]
  else:
    trimy = int((img_rgb.shape[1] - img_rgb.shape[0]) / 2)
    x1 = trimy
    x2 = img_rgb.shape[1] - trimy
    y1 = 0
    y2 = img_rgb.shape[0]
trim_img = img_rgb[y1:y2,x1:x2] #[top : bottom, left : right]
print(img_rgb.shape)
print(trim_img.shape)
plt.imshow(trim_img);
plt.show()

mapMax = 1                                           #画像の長い辺に使う地図の枚数
if mapMax == -1:                                     #-1なら変えない
  resize_img = trim_img
else:                                                #長い辺をimgPixMaxに入れる
  if trim_img.shape[1] >= trim_img.shape[0]:
    imgPixMax = trim_img.shape[1]
  else:
    imgPixMax = trim_img.shape[0]
  if imgPixMax > 128:                                #長い辺が128より大きいなら
    mapPixMax = mapMax * 128                         #画像の長い辺に使う地図の枚数 * 128
    #長い辺に合わせて短い辺を指定
    if trim_img.shape[1] >= trim_img.shape[0]:
      a = (trim_img.shape[1] / mapPixMax)
      mapX = mapPixMax
      mapY = int(trim_img.shape[0] / a)
    else:
      a = (trim_img.shape[0] / mapPixMax)
      mapY = mapPixMax
      mapX = int(trim_img.shape[1] / a)
  else:                                              #長い辺が128より短いなら
    #長い辺に合わせて短い辺を指定
    if trim_img.shape[1] >= trim_img.shape[0]:
      a = 128 / trim_img.shape[1]
      mapX = 128
      mapY = int(trim_img.shape[0] * a)
    else:
      a = 128 / trim_img.shape[1]
      mapY = 128
      mapX = int(trim_img.shape[1] * a)
  resize_img = cv2.resize(trim_img,(mapX,mapY))       #サイズ変換
print(trim_img.shape)
print(resize_img.shape)
plt.imshow(resize_img)
plt.show()
```

```python
color_light = ['-1',' 0','+1']

# 出力用画像の初期化
img_aft = np.zeros_like(resize_img)

iro_how = np.zeros(int(iro.shape[0]/iro.shape[1])+1)

iro_csv = [[' ' for i in range(resize_img.shape[0])] for j in range(resize_img.shape[1])]

block_color = np.zeros((int(resize_img.shape[0]), resize_img.shape[1]))

block_water = np.zeros((int(resize_img.shape[0]), resize_img.shape[1]))

block_light = np.zeros((int(resize_img.shape[0]), resize_img.shape[1]))

# ピクセルごとの色を変換
for x in range(resize_img.shape[0]):
        for y in range(resize_img.shape[1]):
            # 初期化：最初の色との距離
            dx = resize_img[x][y][0] - iro[0][0]
            dy = resize_img[x][y][1] - iro[0][1]
            dz = resize_img[x][y][2] - iro[0][2]
            irosyutu = dx * dx + dy * dy + dz * dz
            ironum = 0

            # iro配列内の各色と比較
            for j in range(iro.shape[0]):
                dx = resize_img[x][y][0] - iro[j][0]
                dy = resize_img[x][y][1] - iro[j][1]
                dz = resize_img[x][y][2] - iro[j][2]
                irohikaku = dx * dx + dy * dy + dz * dz
                if irohikaku < irosyutu:
                    irosyutu = irohikaku
                    ironum = j

            # 一番近い色を出力画像に反映
            img_aft[x][y][0] = iro[ironum][0]
            img_aft[x][y][1] = iro[ironum][1]
            img_aft[x][y][2] = iro[ironum][2]

            a = (ironum//iro.shape[1])
            iro_how[a] = iro_how[a] + 1

            #iro_csv[x][y] = f"{a+1} {color_light[ironum % iro.shape[1]]}"
            iro_csv[x][y] = f"{color_light[ironum % iro.shape[1]]} { block_name[a+1]}"

            block_color[x][y] = int(ironum // iro.shape[1])+1

            if a+1 == 12:
              block_water[x][y] = 1
            else:
              block_water[x][y] = 0

            block_light[x][y] = ironum % iro.shape[1]

for i in range(iro_how.shape[0]-1):
    print(f"{i+1} :{int(iro_how[i])}個")
```
